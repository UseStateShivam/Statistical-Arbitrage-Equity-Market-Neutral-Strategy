# 01: Data Quality Assurance & Point-in-Time Universe Selection

**Objective**: Ingest historical market data, audit schema and pricing integrity, and construct point-in-time eligible equity universes without look-ahead bias.

```
from statarb.config import AppConfig
from statarb.data.ingestion import IngestionManager
from statarb.data.universe import UniverseFilter
import pandas as pd

cfg = AppConfig.from_yaml('../configs/default.yaml')
mgr = IngestionManager(cfg)
df, report = mgr.run_ingestion()
print(f'Total observations: {len(df):,}')
print(f'Symbols audited: {report.total_symbols_evaluated}, Valid: {report.valid_symbols_count}')
```

### Point-in-Time Liquidity & Price Filters
Evaluating equities against minimum price ($5.00) and median dollar volume ($5M) thresholds.

```
filter_engine = UniverseFilter(cfg.universe)
latest_date = df['date'].max()
audit_records = filter_engine.filter_universe(df, as_of_date=latest_date)
audit_df = pd.DataFrame([r.__dict__ for r in audit_records])
audit_df.head(10)
```
