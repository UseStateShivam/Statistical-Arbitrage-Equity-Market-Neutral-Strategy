# 04: Chronological Walk-Forward Backtesting Engine

**Objective**: Execute rolling out-of-sample forward trading with next-session Open fills, 5 basis points slippage per traded side, commissions, short borrow fees, and double-entry accounting.

```
from statarb.config import AppConfig
from statarb.backtest.walk_forward import WalkForwardEngine
from statarb.reporting.tables import format_performance_summary_table
from statarb.metrics.performance import compute_cagr, compute_cumulative_return, compute_annualized_volatility
from statarb.metrics.risk import compute_sharpe_ratio, compute_max_drawdown
import pandas as pd

cfg = AppConfig.from_yaml('../configs/default.yaml')
df = pd.read_parquet('../data/processed/prices.parquet')
engine = WalkForwardEngine(cfg)
wf_res = engine.run(df)

daily_port = wf_res.daily_portfolio
equity = daily_port['portfolio_equity']
returns = daily_port['daily_net_return']
print(f'Out-of-sample CAGR: {compute_cagr(equity):.2%}')
print(f'Sharpe Ratio: {compute_sharpe_ratio(returns):.2f}')
print(f'Max Drawdown: {compute_max_drawdown(equity)[0]:.2%}')
```
